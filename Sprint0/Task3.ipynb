### Вариант задания № 1
**Описание задачи:**
Создать базовый класс `Person` в C#, который будет представлять информацию о человеке. На основе этого класса разработать 3 производных класса (`Student`, `Employee`, `Teacher`), демонстрирующих принципы ООП: инкапсуляцию, наследование, полиморфизм и абстракцию.

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

В данном проекте реализована иерархия классов для предметной области управления людьми в организации/учебном заведении:

1. **Инкапсуляция**: Все ключевые поля классов сокрыты и доступны через свойства с автореализацией, обеспечивая контролируемый доступ к данным.
2. **Абстракция и Наследование**: Создан базовый класс `Person`, от которого наследуются производные классы `Student`, `Employee` и `Teacher`, расширяющие базовую функциональность уникальными атрибутами (`University`, `Company`, `Subject` и др.).
3. **Полиморфизм**: Методы `GetInfo()` и `SayHello()` объявлены как `virtual` в базовом классе и переопределены с помощью ключа `override` в дочерних классах для демонстрации индивидуального поведения каждого типа объекта.

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }

    public Person(string name, int age, string gender)
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }
}

public class Student : Person
{
    public string University { get; set; }
    public int Course { get; set; }

    public Student(string name, int age, string gender, string university, int course)
        : base(name, age, gender)
    {
        University = university;
        Course = course;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я учусь на {Course} курсе в {University}.");
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, ВУЗ: {University}, Курс: {Course}";
    }
}

public class Employee : Person
{
    public string Company { get; set; }
    public decimal Salary { get; set; }

    public Employee(string name, int age, string gender, string company, decimal salary)
        : base(name, age, gender)
    {
        Company = company;
        Salary = salary;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Компания: {Company}, Зарплата: {Salary:C}";
    }
}

public class Teacher : Person
{
    public string Subject { get; set; }
    public int Experience { get; set; }

    public Teacher(string name, int age, string gender, string subject, int experience)
        : base(name, age, gender)
    {
        Subject = subject;
        Experience = experience;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Меня зовут {Name}, я преподаю {Subject} (Стаж: {Experience} лет).");
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Предмет: {Subject}, Стаж: {Experience} лет";
    }
}

List<Person> people = new List<Person>
{
    new Person("Алексей", 30, "Мужской"),
    new Student("Иван", 20, "Мужской", "ТюмГУ", 2),
    new Employee("Ольга", 28, "Женский", "Яндекс", 150000),
    new Teacher("Мария Ивановна", 45, "Женский", "Объектно-ориентированное программирование", 15)
};

Console.WriteLine("=== ДЕМОНСТРАЦИЯ РАБОТЫ МЕТОДОВ SAYHELLO() ===\n");
foreach (var person in people)
{
    person.SayHello();
}

Console.WriteLine("\n=== ДЕМОНСТРАЦИЯ РАБОТЫ МЕТОДОВ GETINFO() ===\n");
foreach (var person in people)
{
    Console.WriteLine(person.GetInfo());
}
```
